# Multi-column GROUP BY

```python
# Q18 — Market + category
q = """ SELECT TRIM(market) as market, TRIM(category) as category, SUM(sales) as total_sales
        FROM ecommerce
        GROUP BY TRIM(market), TRIM(category)
        ORDER BY total_sales DESC;
        """
pd.read_sql_query(q,ecommerce_database)

# Which market-category combination generates the highest sales => market = EU, category = OFFICE SUPPLIES, total sales = 5899758
```

```python
# Q19 — Region + category
q = """select region, category, sum(sales) as total_sales
        from ecommerce
        group by region, category
        order by total_sales desc;"""
pd.read_sql_query(q,ecommerce_database)
```

# 6. HAVING

## Difference between where and having clause
**WHERE → filters individual rows**
**HAVING → filters groups after GROUP BY**

```python
# Q20 — Customers whose total sales exceed 50,000
q = """select customer_name, sum(sales) as total_sales
        from ecommerce
        group by customer_name
        having (sum(sales) > 50000)
        order by total_sales desc;"""
pd.read_sql_query(q,ecommerce_database)
```

```python
# Q.21 Categories with sales greater than 10 million
q = """select category, sum(sales) as total_sales
        from ecommerce
        group by category
        having sum(sales) >10000000
        order by total_sales desc;"""
pd.read_sql_query(q,ecommerce_database)
```

# CASE WHEN

```python
#Q. 22 Classify orders by sales
q = """SELECT
    order_id,
    sales,
    CASE
        WHEN sales >= 1000 THEN 'High'
        WHEN sales >= 500 THEN 'Medium'
        ELSE 'Low'
    END AS sales_category
FROM ecommerce
LIMIT 20;"""
pd.read_sql_query(q,ecommerce_database)
```

```python
# Q23 — Classify discounts
q = """SELECT
    order_id,
    discount,
    CASE
        WHEN discount >= 0.30 THEN 'High Discount'
        WHEN discount >= 0.10 THEN 'Medium Discount'
        ELSE 'Low Discount'
    END AS discount_category
FROM ecommerce
LIMIT 20;"""
pd.read_sql_query(q,ecommerce_database)
```

```python
# Q24 — Sales performance by market
q = """
    SELECT 
        TRIM(market) AS market,
        COUNT(DISTINCT order_id) AS total_orders,
        SUM(quantity) AS total_quantity,
        SUM(sales) AS total_sales,
        AVG(sales) AS avg_transaction_sales
    FROM ecommerce
    GROUP BY TRIM(market)
    ORDER BY total_sales DESC;
"""
pd.read_sql_query(q,ecommerce_database)
```

```python
q = """SELECT
    typeof(sales) AS stored_type,
    COUNT(*) AS row_count
FROM ecommerce
GROUP BY typeof(sales);"""
pd.read_sql_query(q,ecommerce_database)
```

```python
# Q.3 Calculate sales by year
# Use strftime('%Y', order_date) if the dates are stored as SQLite dates.

q = """
SELECT 
    strftime('%Y', order_date) AS sales_year,
    SUM(sales) AS total_sales
FROM 
    ecommerce
GROUP BY 
    sales_year
ORDER BY 
    total_sales DESC;
"""


pd.read_sql_query(q, ecommerce_database)
```
