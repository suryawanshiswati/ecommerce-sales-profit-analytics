## Part 2 — Data Quality & Cleaning

```python
# Missing values
missing_values = data.isna().sum().sort_values(ascending=False)
missing_values[missing_values > 0]
```

```python
# Exact duplicate rows
print('Exact duplicate rows:', data.duplicated().sum())
```

```python
# Inspect duplicate rows before deciding whether to remove them
duplicates = data[data.duplicated(keep=False)].sort_values(list(data.columns))
print('Rows involved in exact duplicates:', len(duplicates))
duplicates.head(40)
```

### Product variant column

`product_name.1` is not a duplicate of `product_name`; it contains product attributes such as Blue, Laser, Wireless, Premium, etc. Therefore it is retained and renamed to `product_variant`. Missing variants are not a reason to delete the transaction rows.

```python
# Rename the product attribute column
data.rename(columns={'product_name.1': 'product_variant'}, inplace=True)

# Clean text and explicitly label missing variants
data['product_variant'] = (
    data['product_variant']
    .fillna('No Variant Specified')
    .astype(str)
    .str.strip()
)

print('Missing product_variant values:', data['product_variant'].isna().sum())
print('Unique product variants:', data['product_variant'].nunique())
data['product_variant'].value_counts().head(20)
```

### Missing identifiers

There are 21 rows with a missing `order_id` or `product_id`. We will **not delete them from the raw dataset**, because they still contain sales and financial information. Instead, analyses requiring a valid order/product identifier should exclude those rows explicitly.

```python
missing_id_rows = data[data['order_id'].isna() | data['product_id'].isna()]
print('Rows with missing order_id or product_id:', len(missing_id_rows))
missing_id_rows[['order_id', 'product_id', 'product_name', 'sales']].head(25)
```

### Validate the year field

The raw `year` column contains an invalid value (2088) and does not always agree with `order_date`. For analysis, we will create `order_year` directly from the trusted `order_date` field rather than silently overwriting the source column.

```python
data['order_year'] = data['order_date'].dt.year

year_mismatches = data[data['year'] != data['order_year']]
print('Rows where year differs from order_date year:', len(year_mismatches))
year_mismatches[['order_date', 'year', 'order_year']].head(20)
```

```python
print('Raw year range:', data['year'].min(), 'to', data['year'].max())
print('Validated order_year range:', data['order_year'].min(), 'to', data['order_year'].max())
```

```python
# Use order_year for time-based analysis
print(data['order_year'].value_counts().sort_index())
```

## Part 3 — Validate Financial Columns

```python
# Exploratory calculated profit — this is NOT assumed to be an official dataset field yet
data['calculated_profit'] = data['revenue'] - data['expenses']

# Check how revenue relates to sales
data['revenue_minus_sales'] = data['revenue'] - data['sales']

print('Total Sales:', data['sales'].sum())
print('Total Expenses:', data['expenses'].sum())
print('Total Revenue:', data['revenue'].sum())
print('Calculated Profit (Revenue - Expenses):', data['calculated_profit'].sum())
```

**Important:** We use `calculated_profit = revenue - expenses` as an exploratory metric for this project. We should not claim it is the dataset creator’s official definition of profit unless the source documentation confirms it.

## Part 4 — Initial Business Analysis

### Sales KPIs

```python
total_sales = data['sales'].sum()
unique_orders = data['order_id'].nunique()

# AOV must be based on unique orders, not number of transaction rows
aov = total_sales / unique_orders

print(f'Total Sales: {total_sales:,.2f}')
print(f'Unique Orders: {unique_orders:,}')
print(f'Average Order Value (AOV): {aov:,.2f}')
```

```python
# Average sales per transaction row — useful, but NOT AOV
avg_transaction_sales = data['sales'].mean()
print(f'Average Sales per Transaction Row: {avg_transaction_sales:,.2f}')
```

### Q1–Q5: Sales performance

```python
# Q1. Total sales
print('Q1 — Total Sales:', f"{data['sales'].sum():,.2f}")

# Q2. Average Order Value
print('Q2 — Average Order Value:', f"{data['sales'].sum() / data['order_id'].nunique():,.2f}")

# Q3. Highest-sales category
category_sales = data.groupby('category')['sales'].sum().sort_values(ascending=False)
print('Q3 — Highest-sales category:', category_sales.idxmax())

# Q4. Highest-sales sub-category
subcategory_sales = data.groupby('sub_category')['sales'].sum().sort_values(ascending=False)
print('Q4 — Highest-sales sub-category:', subcategory_sales.idxmax())

# Q5. Highest-sales market
market_sales = data.groupby('market')['sales'].sum().sort_values(ascending=False)
print('Q5 — Highest-sales market:', market_sales.idxmax())
```

### Q6–Q7: Customer performance

```python
segment_sales = data.groupby('segment')['sales'].sum().sort_values(ascending=False)
print('Q6 — Highest-sales customer segment:', segment_sales.idxmax())
print(segment_sales)

print('\nQ7 — Top 10 customers by sales:')
top_10_customers = data.groupby('customer_name')['sales'].sum().sort_values(ascending=False).head(10)
print(top_10_customers)
```

### Q8: Product performance

```python
top_10_products = data.groupby('product_name')['sales'].sum().sort_values(ascending=False).head(10)
print('Q8 — Top 10 products by sales:')
print(top_10_products)
```

### Q9–Q10: Geographic performance

```python
top_countries = data.groupby('country')['sales'].sum().sort_values(ascending=False).head(10)
print('Q9 — Top 10 countries by sales:')
print(top_countries)

region_sales = data.groupby('region')['sales'].sum().sort_values(ascending=False)
print('\nQ10 — Sales by region:')
print(region_sales)
```

## Data-quality decisions made 

1. Standardized column names by stripping whitespace.
2. Retained `product_name.1`, renamed it to `product_variant`, and labeled missing variants as `No Variant Specified`.
3. Preserved the 21 rows with missing identifiers instead of deleting transaction data.
4. Inspected exact duplicates before deciding whether to remove them.
5. Created `order_year` from `order_date` because the raw `year` field contains inconsistencies.
6. Corrected AOV to use total sales divided by unique orders.
7. Kept `calculated_profit` explicitly as an exploratory metric rather than assuming it is the official profit definition.
